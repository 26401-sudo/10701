```python
import random

com = random.randint(1, 10)
num = 0

while num != com:
    num = int(input('1에서 10 사이의 수를 입력하세요: '))

    if num < com:
        print('입력한 값이 더 작습니다.')
    elif num > com:
        print('입력한 값이 더 큽니다.')
print('정답입니다!')
```
