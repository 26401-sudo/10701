```python
n = int(input("2 이상의 정수를 입력하세요: "))

count = 0

for i in range(1, n + 1):
    if n % i == 0:
        count = count + 1
if count == 2:
    print(n, "은(는) 소수입니다.")
else:
    print(n, "은(는) 소수가 아닙니다.")
```
